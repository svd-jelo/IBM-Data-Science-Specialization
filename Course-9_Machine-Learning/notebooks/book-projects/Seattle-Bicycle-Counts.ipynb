```python
import pandas as pd
```

```python
def update_data(data, api_endpoint, api_limit=1000):
    """
    Returns updated dataset given as a pandas Dataframe based on given API endpoint and API limit.
    Note: it is assumed that updated and outdated datasets have the same set of columns
    
    data: Dataframe
        Dataframe containing outdated dataset
        
    api_endpoint: str
        Link to API Endpoint for updated dataset
        
    api_limit: int, default 1000
        Default API limit for exporting updated dataset from API Endpoint
    """
    offset = data.shape[0]
    path = api_endpoint + '?$offset={}'
    data_updated_list = [data, pd.read_csv(path.format(offset), header=0, names=data.columns)]
    
    if data_updated_list[-1].shape[0] == 0:
        return data
        
    while data_updated_list[-1].shape[0] >= 1000:
        offset += api_limit
        data_updated_list.append(
            pd.read_csv(path.format(offset), header=0, names=data.columns)
        )

    return pd.concat(data_updated_list, ignore_index=True)
```

```python
csv_path = '~/IBM-Data-Science-Specialization/Course-9_Machine-Learning/data/Fremont_Bridge_Bicycle_Counter_20260527.csv'
#data = pd.read_csv(csv_path, index_col='Date', parse_dates=True, date_format='%m/%d/%Y %I:%M:%S %p')
#data.columns = ['Total', 'West', 'East']
data_old = pd.read_csv(csv_path, nrows=117000)
api_endpoint = 'https://data.seattle.gov/resource/65db-xm6k.csv'
data_old.shape
```

```python
data = update_data(data_old, api_endpoint)
data.shape
```

```python
data['Date'] = pd.to_datetime(data.Date, format='ISO8601')
```

```python
data.set_index('Date', inplace=True)
```

```python
data.columns = ['Total', 'West', 'East']
```
